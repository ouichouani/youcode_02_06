```python
import joblib
data = joblib.load("../data/02_data.joblib")
x_train = data["x_train"] 
x_test = data["x_test"] 
y_train = data["y_train"] 
y_test = data["y_test"]
```

```python
from xgboost import XGBClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC

# XGBoost
model_XGBClassifier = XGBClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=6,
    random_state=100,
    eval_metric='mlogloss'
)

# Random Forest
model_RandomForestClassifier = RandomForestClassifier(
    n_estimators=100,
    criterion='gini',
    max_depth=None,
    min_samples_split=2,
    min_samples_leaf=1,
    max_features='sqrt',
    bootstrap=True,
    class_weight=None,
    random_state=100,
    n_jobs=-1
)

# svm
model_SVC = SVC(
    C=1,
    kernel='rbf',
    gamma='scale',
    class_weight='balanced',
    probability=True,
    random_state=100
)

models = {
    "XGBClassifier" : model_XGBClassifier ,
    "RandomForestClassifier" : model_RandomForestClassifier ,
    "svm" : model_SVC ,
}
```

```python
for name, model in models.items():
    print(f"Training {name}...")
    model.fit(x_train, y_train)
    print(f"{name} trained successfully!")

y_preds = {}
for name , model in models.items() :
    y_preds[name] = model.predict(x_test)
```

```python
import pandas as pd
from sklearn.metrics import (
    accuracy_score ,
    precision_score ,
    recall_score , 
    f1_score ,
    confusion_matrix ,
    classification_report ,
    roc_auc_score
)

results = []
for name , model in models.items() :
    y_proba = model.predict_proba(x_test)

    model_accuracy_score = accuracy_score(y_test , y_pred=y_preds[name] )
    model_precision_score = precision_score(y_test , y_pred=y_preds[name] , average='macro' , zero_division=0)
    model_recall_score = recall_score(y_test , y_pred=y_preds[name] , average='macro' , zero_division=0)
    model_f1_score = f1_score(y_test , y_pred=y_preds[name] , average='macro' , zero_division=0)
    model_confusion_matrice = confusion_matrix(y_test, y_pred=y_preds[name])
    model_classification_report = classification_report(y_test, y_preds[name], zero_division=0 , output_dict=True,)
    model_roc_auc_score = roc_auc_score(y_test , y_proba ,multi_class="ovr",average="macro")


    results.append({
        "Model": name,
        "Accuracy":model_accuracy_score ,
        "Precision": model_precision_score , 
        "Recall": model_recall_score ,
        "F1-Score": model_f1_score , 
        "ROC-AUC": model_roc_auc_score
    })

    report = pd.DataFrame(model_classification_report)
    results_df = pd.DataFrame(results)

display(report)
display(results_df)
```
